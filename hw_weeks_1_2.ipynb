**Réponse théorique** :

On cherche les $z \in \mathbb{C}$ tels que $z^3 = -8$.

**1. Forme polaire du second membre**

$$-8 = 8\,e^{j\pi}$$

(module $8$, argument $\pi$).

**2. Forme polaire de $z$**

On pose $z = r\,e^{j\theta}$, donc $z^3 = r^3 e^{j3\theta}$. L'égalité $r^3 e^{j3\theta} = 8\,e^{j\pi}$ impose :

- Modules : $r^3 = 8 \Rightarrow r = 2$
- Arguments : $3\theta = \pi + 2k\pi \Rightarrow \theta_k = \dfrac{\pi + 2k\pi}{3}, \quad k \in \{0,1,2\}$

**3. Les trois solutions**

| $k$ | Forme polaire | Forme cartésienne |
|---|---|---|
| 0 | $z_1 = 2\,e^{j\pi/3}$ | $z_1 = 2\left(\cos\frac{\pi}{3} + j\sin\frac{\pi}{3}\right) = 1 + j\sqrt{3}$ |
| 1 | $z_2 = 2\,e^{j\pi}$ | $z_2 = 2(\cos\pi + j\sin\pi) = -2$ |
| 2 | $z_3 = 2\,e^{j5\pi/3}$ | $z_3 = 2\left(\cos\frac{5\pi}{3} + j\sin\frac{5\pi}{3}\right) = 1 - j\sqrt{3}$ |

**4. Vérification théorique**

$$z_1^3 = 8\,e^{j\pi} = -8, \quad z_2^3 = (-2)^3 = -8, \quad z_3^3 = 8\,e^{j5\pi} = 8\,e^{j\pi} = -8 \;\checkmark$$

**Interprétation géométrique** : les trois racines sont sur le cercle de rayon $2$ et forment un triangle équilatéral (angles espacés de $2\pi/3$). Les racines $z_1$ et $z_3$ sont conjuguées, car l'équation est à coefficients réels.

### Exercice 2 – Lien entre partie réelle et imaginaire
Soit $z = e^{j\theta}$. Montrer que
$$ \cos\theta = \frac{z+z^{-1}}{2} \quad\text{et}\quad \sin\theta = \frac{z-z^{-1}}{2j}. $$
En déduire une expression de $\cos^3\theta$ en fonction de $\cos(3\theta)$ et $\cos\theta$ (formule de triple angle). Utiliser cette relation pour exprimer $\cos^3\theta$ comme combinaison linéaire de cosinus.

```python
# Vérification numérique de la formule
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 100)
cos3 = np.cos(theta)**3

# Formule dérivée : cos^3 = (3/4) cos(theta) + (1/4) cos(3 theta)
cos3_formule = (3/4)*np.cos(theta) + (1/4)*np.cos(3*theta)

plt.plot(theta, cos3, label='cos^3')
plt.plot(theta, cos3_formule, '--', label='formule : 3/4 cos(θ) + 1/4 cos(3θ)')
plt.xlabel('θ (rad)')
plt.legend()
plt.grid()
plt.show()

# Erreur maximale entre les deux courbes
print("Erreur max :", np.max(np.abs(cos3 - cos3_formule)))
print("Formules identiques :", np.allclose(cos3, cos3_formule))

# Vérification de la formule de triple angle : cos(3θ) = 4cos^3 - 3cos
print("Triple angle OK :",
      np.allclose(np.cos(3*theta), 4*np.cos(theta)**3 - 3*np.cos(theta)))
```

**Démonstration** :

**1. Expressions de $\cos\theta$ et $\sin\theta$**

Soit $z = e^{j\theta}$, donc $z^{-1} = e^{-j\theta}$. D'après la formule d'Euler :

$$e^{j\theta} = \cos\theta + j\sin\theta, \qquad e^{-j\theta} = \cos\theta - j\sin\theta$$

- En additionnant : $z + z^{-1} = 2\cos\theta \Rightarrow \cos\theta = \dfrac{z+z^{-1}}{2}$
- En soustrayant : $z - z^{-1} = 2j\sin\theta \Rightarrow \sin\theta = \dfrac{z-z^{-1}}{2j}$

**2. Expression de $\cos^3\theta$**

$$\cos^3\theta = \left(\frac{z+z^{-1}}{2}\right)^3 = \frac{1}{8}\left(z^3 + 3z^2 z^{-1} + 3 z z^{-2} + z^{-3}\right) = \frac{1}{8}\left(z^3 + z^{-3} + 3(z + z^{-1})\right)$$

Or $z^3 + z^{-3} = e^{j3\theta} + e^{-j3\theta} = 2\cos 3\theta$ et $z + z^{-1} = 2\cos\theta$. Donc :

$$\cos^3\theta = \frac{1}{8}\left(2\cos 3\theta + 6\cos\theta\right)$$

$$\boxed{\cos^3\theta = \frac{3}{4}\cos\theta + \frac{1}{4}\cos 3\theta}$$

**3. Formule de triple angle**

En isolant $\cos 3\theta$ :

$$\cos 3\theta = 4\cos^3\theta - 3\cos\theta$$

**Conclusion** : $\cos^3\theta$ est une combinaison linéaire de cosinus de fréquences $\theta$ et $3\theta$ (on a « linéarisé » la puissance).

### Exercice 3 – Équation avec exponentielle
Trouver tous les réels $\theta$ tels que $e^{j\theta} + e^{-j\theta} = \sqrt{3}$.

Que valent $\cos\theta$ et $\sin\theta$ ?

**Solution** (à compléter) :

**Solution** :

**1. Simplification**

D'après la formule d'Euler :

$$e^{j\theta} + e^{-j\theta} = 2\cos\theta$$

L'équation devient donc :

$$2\cos\theta = \sqrt{3} \iff \cos\theta = \frac{\sqrt{3}}{2}$$

**2. Résolution dans $\mathbb{R}$**

On sait que $\cos\dfrac{\pi}{6} = \dfrac{\sqrt{3}}{2}$. Le cosinus étant pair et $2\pi$-périodique :

$$\boxed{\theta = \pm\frac{\pi}{6} + 2k\pi, \quad k \in \mathbb{Z}}$$

**3. Valeurs de $\cos\theta$ et $\sin\theta$**

On a $\cos\theta = \dfrac{\sqrt{3}}{2}$ dans tous les cas. Pour le sinus, on utilise $\sin^2\theta = 1 - \cos^2\theta = 1 - \dfrac{3}{4} = \dfrac{1}{4}$, donc $\sin\theta = \pm\dfrac{1}{2}$, avec :

| $\theta$ | $\cos\theta$ | $\sin\theta$ |
|---|---|---|
| $\dfrac{\pi}{6} + 2k\pi$ | $\dfrac{\sqrt{3}}{2}$ | $+\dfrac{1}{2}$ |
| $-\dfrac{\pi}{6} + 2k\pi$ | $\dfrac{\sqrt{3}}{2}$ | $-\dfrac{1}{2}$ |

Le sinus n'est donc pas déterminé de façon unique par l'équation : seul son signe change selon la solution.

**Interprétation** : $e^{j\theta} + e^{-j\theta}$ est la somme de deux nombres complexes conjugués, donc un réel égal à deux fois la partie réelle.

```python
# Génération du signal
import numpy as np
import matplotlib.pyplot as plt

fs = 10000  # fréquence d'échantillonnage (Hz)
t = np.arange(0, 0.02, 1/fs)

x = (np.cos(2*np.pi*100*t)
     + 2*np.cos(2*np.pi*200*t - np.pi/4)
     + 0.5*np.cos(2*np.pi*300*t + np.pi/3))

# Version complexe avec phaseurs
X1, X2, X3 = 1, 2*np.exp(-1j*np.pi/4), 0.5*np.exp(1j*np.pi/3)
x_complexe = np.real(X1*np.exp(1j*2*np.pi*100*t)
                     + X2*np.exp(1j*2*np.pi*200*t)
                     + X3*np.exp(1j*2*np.pi*300*t))
print("Forme réelle = forme complexe :", np.allclose(x, x_complexe))

plt.figure(figsize=(10, 4))
plt.plot(t, x)
plt.axvline(0.01, color='r', linestyle='--', label='T0 = 10 ms')
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.title('Superposition de trois sinusoïdes')
plt.legend()
plt.grid()
plt.show()
```

**Réponse** :

**1. Chaque terme sous forme d'exponentielle complexe** (avec $\cos(\alpha) = \operatorname{Re}\{e^{j\alpha}\}$)

- $\cos(2\pi\,100\,t) = \operatorname{Re}\{e^{j2\pi 100 t}\}$
- $2\cos(2\pi\,200\,t - \pi/4) = \operatorname{Re}\{2e^{-j\pi/4}\,e^{j2\pi 200 t}\}$
- $0{,}5\cos(2\pi\,300\,t + \pi/3) = \operatorname{Re}\{0{,}5e^{j\pi/3}\,e^{j2\pi 300 t}\}$

**2. Expression complexe complète**

$$x(t) = \operatorname{Re}\left\{ e^{j2\pi 100 t} + 2e^{-j\pi/4}e^{j2\pi 200 t} + 0{,}5e^{j\pi/3}e^{j2\pi 300 t} \right\}$$

Les phaseurs (amplitude complexe) sont $X_1 = 1$, $X_2 = 2e^{-j\pi/4}$, $X_3 = 0{,}5e^{j\pi/3}$.

**3. Observation**

Les fréquences 100, 200 et 300 Hz sont des multiples entiers de 100 Hz : ce sont la fondamentale et ses harmoniques 2 et 3. Le signal est donc **périodique de période $T_0 = 1/100 = 10$ ms** (deux périodes visibles sur 20 ms). Sa forme n'est **pas sinusoïdale** : la superposition produit une onde complexe, dont les sommets sont dissymétriques à cause des phases différentes. L'amplitude maximale est au plus $1 + 2 + 0{,}5 = 3{,}5$.

Deux signaux de fréquences proches :  
$
x_1(t) = \cos(2\pi f_1 t)
$ et 
$
x_2(t) = \cos(2\pi f_2 t)
$.  
Montrer que leur somme peut s’écrire  
$$ 
2\cos(2\pi f_m t)\cos(2\pi f_d t) 
$$  
avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$.  
Pour $f_1 = 10$ Hz et $f_2 = 12$ Hz, tracer la somme et observer l’enveloppe. Expliquer le phénomène de battement.

**Démonstration** (à compléter) :

```python
f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1+f2)/2
fd = (f1-f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)  # enveloppe
x_product = x_env * np.cos(2*np.pi*fm*t)

plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

# Explication : la somme de deux sinusoïdes de fréquences proches donne une oscillation rapide (fm) modulée par une enveloppe lente (fd).
```

```python
import numpy as np
import matplotlib.pyplot as plt

f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2*np.pi*f1*t)
x2 = np.cos(2*np.pi*f2*t)
x_sum = x1 + x2

fm = (f1 + f2)/2
fd = (f1 - f2)/2
x_env = 2*np.cos(2*np.pi*fd*t)           # enveloppe
x_product = x_env * np.cos(2*np.pi*fm*t)

plt.figure(figsize=(12, 5))
plt.plot(t, x_sum, label='Somme')
plt.plot(t, x_env, '--', label='Enveloppe')
plt.plot(t, -x_env, '--', label='-Enveloppe')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

print("Somme = produit 2cos(2π fm t)cos(2π fd t) :", np.allclose(x_sum, x_product))
print(f"Fréquence de battement : {abs(f1 - f2)} Hz")

# Explication : la somme de deux sinusoïdes de fréquences proches donne une
# oscillation rapide (fm) modulée par une enveloppe lente (fd).
```

### Exercice 8 – Modulation d’amplitude
Un signal modulé en amplitude s’écrit  
$$ 
s(t) = A_c \big[1 + m \cos(2\pi f_m t)\big] \cos(2\pi f_c t).
$$  
Générer ce signal pour $f_c = 100$ Hz, $f_m = 5$ Hz, $m = 0.5$, $A_c = 1$.  
Tracer le signal sur 0,2 s, puis calculer sa transformée de Fourier (FFT) pour visualiser les raies spectrales (porteuse et bandes latérales).  
Expliquer pourquoi il y a trois composantes fréquentielles.

**Code** :

```python
fs = 1000  # Hz
t = np.linspace(0, 0.2, int(0.2*fs))
fc = 100
fm = 5
m = 0.5
Ac = 1

s = Ac * (1 + m*np.cos(2*np.pi*fm*t)) * np.cos(2*np.pi*fc*t)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(t, s)
plt.xlabel('Temps (s)')
plt.title('Signal modulé en amplitude')
plt.grid()

# FFT
N = len(s)
freqs = np.fft.fftfreq(N, 1/fs)
spectre = np.fft.fft(s)

plt.subplot(1, 2, 2)
plt.stem(freqs[:N//2], np.abs(spectre[:N//2]) / N)
plt.xlim(0, 150)
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Amplitude')
plt.title('Spectre (3 raies)')
plt.grid()
plt.show()

# Explication : le produit de cosinus donne des composantes à fc, fc+fm, fc-fm.
```

**Réponse** :

**Pourquoi trois composantes ?**

On développe le produit avec $\cos a\cos b = \frac{1}{2}[\cos(a+b) + \cos(a-b)]$ :

$$s(t) = A_c\cos(2\pi f_c t) + \frac{A_c m}{2}\cos\bigl(2\pi (f_c+f_m) t\bigr) + \frac{A_c m}{2}\cos\bigl(2\pi (f_c-f_m) t\bigr)$$

Le signal est la somme de trois sinusoïdes pures :

| Composante | Fréquence | Amplitude |
|---|---|---|
| Porteuse | $f_c = 100$ Hz | $A_c = 1$ |
| Bande latérale supérieure | $f_c + f_m = 105$ Hz | $A_c m/2 = 0{,}25$ |
| Bande latérale inférieure | $f_c - f_m = 95$ Hz | $A_c m/2 = 0{,}25$ |

Le spectre a donc trois raies. L'information (le message à $f_m$) se trouve dans les bandes latérales, de part et d'autre de la porteuse.

### Exercice 10 – Synthèse d’un signal carré par séries de Fourier
Un signal carré de période $T$ et d’amplitude 1 peut être décomposé en série de Fourier :  
$$
x(t) = \frac{4}{\pi} \sum_{k=0}^{N} \frac{1}{2k+1} \sin\bigl(2\pi (2k+1) f_0 t\bigr).
$$  
Écrire une fonction Python qui génère ce signal pour un nombre d’harmoniques $N$ donné.  
Tester pour $N=1, 5, 20, 50$ et tracer les signaux correspondants.  
Observer le phénomène de Gibbs (oscillations près des discontinuités).  
Comparer avec la fonction `square` de `scipy.signal`.

**Code** :

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def carre_fourier(t, f0, N):
    """Génère un signal carré par synthèse de Fourier avec N harmoniques impaires."""
    x = np.zeros_like(t)
    for k in range(N):
        n = 2*k + 1
        x += (1/n) * np.sin(2*np.pi*n*f0*t)
    return (4/np.pi) * x

f0 = 1  # Hz
t = np.linspace(0, 2, 1000)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
Ns = [1, 5, 20, 50]

for ax, N in zip(axes.flat, Ns):
    x = carre_fourier(t, f0, N)
    ax.plot(t, x, label=f'N={N}')
    ax.grid()
    ax.legend()

plt.suptitle('Synthèse d\'un signal carré par séries de Fourier')
plt.tight_layout()
plt.show()

# Comparaison avec square
x_square = signal.square(2*np.pi*f0*t)
plt.plot(t, x_square, label='square (scipy)')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def carre_fourier(t, f0, N):
    """Génère un signal carré par synthèse de Fourier avec N harmoniques impaires."""
    x = np.zeros_like(t)
    for k in range(N):
        n = 2*k + 1                          # harmonique impaire
        x += (1/n) * np.sin(2*np.pi*n*f0*t)
    return (4/np.pi) * x

f0 = 1  # Hz
t = np.linspace(0, 2, 2000)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
Ns = [1, 5, 20, 50]

for ax, N in zip(axes.flat, Ns):
    x = carre_fourier(t, f0, N)
    ax.plot(t, x, label=f'N={N}')
    ax.axhline(1, color='gray', linestyle=':')
    ax.axhline(-1, color='gray', linestyle=':')
    ax.grid()
    ax.legend()
    print(f"N = {N:2d} : maximum = {x.max():.4f}")

plt.suptitle("Synthèse d'un signal carré par séries de Fourier")
plt.tight_layout()
plt.show()

# Comparaison avec square
x_square = signal.square(2*np.pi*f0*t)
plt.figure(figsize=(10, 4))
plt.plot(t, x_square, label='square (scipy)')
plt.plot(t, carre_fourier(t, f0, 50), '--', label='N=50')
plt.legend()
plt.grid()
plt.show()

# Phénomène de Gibbs : le maximum reste ≈ 1.18 quand N augmente (dépassement ≈ 9 % du saut de 2)
```


**Principe** : un signal carré impair, de période $T = 1/f_0$ et d'amplitude 1, ne contient que des harmoniques **impaires** ($f_0, 3f_0, 5f_0, \dots$), d'amplitudes décroissantes en $1/(2k+1)$. Plus $N$ est grand, plus la somme se rapproche du carré.

**Phénomène de Gibbs** : près des discontinuités, la somme partielle présente des oscillations et un **dépassement d'environ 9 % du saut** (le saut vaut 2, donc le maximum atteint environ $1{,}18$). Lorsque $N$ augmente :

- les oscillations se **resserrent** vers la discontinuité ;
- mais la hauteur du dépassement **ne diminue pas** (elle tend vers $\approx 0{,}179$).

La série converge au sens quadratique (en énergie), mais pas uniformément au voisinage des discontinuités.

**Comparaison avec `scipy.signal.square`** : le signal de Fourier à $N=50$ épouse bien le carré idéal sur les paliers, mais oscille aux transitions. `square` donne le carré idéal, avec des transitions verticales.

### Exercice 11 – Phase et décalage temporel
Montrer que $A\cos(2\pi f t - \phi)$ est équivalent à un décalage temporel. Quel est le décalage en secondes ?  
Pour $A=1, f=2$ Hz, $\phi = \pi/2$, calculer le décalage. Tracer le signal non décalé et le signal décalé pour vérifier.

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

**Code** :

```python
f = 2
phi = np.pi/2
A = 1
delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s")

t = np.linspace(0, 1, 1000)
x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

plt.plot(t, x_original, label='Original (phi=0)')
plt.plot(t, x_decale, '--', label='Décalé temporel')
plt.plot(t, x_phase, ':', label='Phase -π/2')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

f = 2
phi = np.pi/2
A = 1
delta_t = phi / (2*np.pi*f)
print(f"Décalage temporel : {delta_t:.4f} s  (T/4 = {1/(4*f):.4f} s)")

t = np.linspace(0, 1, 1000)
x_original = A * np.cos(2*np.pi*f*t)
x_decale = A * np.cos(2*np.pi*f*(t - delta_t))
x_phase = A * np.cos(2*np.pi*f*t - phi)

plt.figure(figsize=(10, 4))
plt.plot(t, x_original, label='Original (phi=0)')
plt.plot(t, x_decale, '--', label='Décalé temporel')
plt.plot(t, x_phase, ':', linewidth=3, label='Phase -π/2')
plt.xlabel('Temps (s)')
plt.legend()
plt.grid()
plt.show()

print("Décalage temporel = déphasage :", np.allclose(x_decale, x_phase))
print("cos(2πft - π/2) = sin(2πft)   :", np.allclose(x_phase, np.sin(2*np.pi*f*t)))
```

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

Ainsi $A\cos(2\pi f t - \phi)$ est le signal $A\cos(2\pi f t)$ **retardé** de $\Delta t$ (décalé vers la droite pour $\phi > 0$). Un décalage temporel se traduit par un déphasage proportionnel à la fréquence.

**Application** : $A=1$, $f=2$ Hz, $\phi = \pi/2$ :

$$\Delta t = \frac{\pi/2}{2\pi\cdot 2} = \frac{1}{8} = 0{,}125 \text{ s}$$

La période est $T = 1/f = 0{,}5$ s, donc $\Delta t = T/4$ : un déphasage de $\pi/2$ correspond à un quart de période. On a aussi $\cos(2\pi f t - \pi/2) = \sin(2\pi f t)$.

### Exercice 12 – Échantillonnage d’une exponentielle complexe
Soit $z[n] = e^{j\omega_0 n}$. Montrer que si $\omega_0$ est un multiple rationnel de $2\pi$, le signal est périodique.  
Pour $\omega_0 = 2\pi/3$, déterminer la période fondamentale.  
Pour $\omega_0 = 1$ rad/échantillon, le signal est-il périodique ? Justifier.  
Générer les deux signaux sur 50 échantillons et observer leur comportement.

**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.

**Code** :

```python
n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(f'$\\omega_0 = 2\pi/3$ (périodique, période 3)')

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title('$\\omega_0 = 1$ (non périodique)')

plt.tight_layout()
plt.show()

# Vérification : pour omega1, la période est 3 car 2pi/3 * 3 = 2pi.
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(r'$\omega_0 = 2\pi/3$ (périodique, période 3)')
plt.xlabel('n')

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title(r'$\omega_0 = 1$ (non périodique)')
plt.xlabel('n')

plt.tight_layout()
plt.show()

# Vérification : pour omega1, la période est 3 car 2pi/3 * 3 = 2pi.
print("z1 périodique de période 3 :", np.allclose(z1[:-3], z1[3:]))

# Pour omega2, on ne trouve aucune période N (sur 1 <= N < 40)
periodes = [N for N in range(1, 40) if np.allclose(z2[:-N], z2[N:])]
print("Périodes trouvées pour z2 :", periodes if periodes else "aucune")
```

**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.

**1. Condition de périodicité**

$z[n]$ est périodique de période $N \in \mathbb{N}^*$ si $z[n+N] = z[n]$ pour tout $n$ :

$$e^{j\omega_0(n+N)} = e^{j\omega_0 n} \iff e^{j\omega_0 N} = 1 \iff \omega_0 N = 2\pi k, \; k\in\mathbb{Z} \iff \frac{\omega_0}{2\pi} = \frac{k}{N}$$

Donc si $\omega_0/(2\pi)$ est rationnel, le signal est périodique. La **période fondamentale** est le plus petit $N$, c'est-à-dire le **dénominateur de la fraction irréductible** $k/N$.

**2. Cas $\omega_0 = 2\pi/3$**

$\dfrac{\omega_0}{2\pi} = \dfrac{1}{3}$, donc la période fondamentale est $N = 3$. On vérifie : $e^{j(2\pi/3)\cdot 3} = e^{j2\pi} = 1$. Le signal prend les 3 valeurs $1,\; e^{j2\pi/3},\; e^{j4\pi/3}$, qui se répètent.

**3. Cas $\omega_0 = 1$ rad/échantillon**

$\dfrac{\omega_0}{2\pi} = \dfrac{1}{2\pi}$ est **irrationnel** (car $\pi$ est irrationnel). Aucun entier $N$ ne vérifie $N = 2\pi k$, donc le signal **n'est pas périodique** : il ne repasse jamais exactement par une valeur déjà prise, même s'il reste borné ($|z[n]|=1$).

**Remarque** : $e^{j\omega_0 t}$ en temps continu est toujours périodique (période $2\pi/\omega_0$). En temps discret, la périodicité n'est possible que si $\omega_0/2\pi$ est rationnel.
